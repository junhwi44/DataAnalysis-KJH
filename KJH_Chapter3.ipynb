**데이터 확인[KJH 코드 3-22]**

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('temp_KJH.csv', encoding='cp949')
df.head(3)
```

**데이터프레임 개요 확인[KJH 코드 3-23]**

```python
df.info()
```

**결측지 채우기[KJH 코드 3-24]**

```python
df2 = df.fillna(method='ffill')
df2.info()
```

**특수 문자가 포함된 열 이름을 변경[KJH 코드 3-25]**

```python
df2.rename(columns={'평균기온(°C)':'avg_temp'}, inplace=True)
df2.rename(columns={'최고기온(°C)':'max_temp'}, inplace=True)
df2.info()
```

**일시 데이터를 날짜형으로 변경[KJH 코드 3-26]**

```python
df2['일시'] = pd.to_datetime(df2['일시'], format='%Y-%m-%d')
df2.info()
```

**일시 데이터를 날짜형으로 변경[KJH 코드 3-27]**

```python
df3 = df2.set_index('일시')
df3.head(3)
```

**9월 데이터 추출[KJH 코드 3-28]**

```python
df_Sep = df3[pd.DatetimeIndex(df3.index).month == 9]
df_Sep
```

**9월 기온 꺾은선 그래프 생성[KJH 코드 3-29]**

```python
plt.title('Seoul 2024. 9')
plt.plot(range(0, 30), df_Sep['max_temp'], label='max', c='r')
plt.plot(range(0, 30), df_Sep['avg_temp'], label='avg', c='y')

plt.xlim(1, 30)
plt.ylim(20, 37)
plt.legend()
plt.rcParams['figure.figsize'] = (8, 8)   # 그림의 가로 세로 사이즈를 설정
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 파일명은 실제 업로드한 이름으로 맞추기
df = pd.read_csv('jeju_2025.csv', encoding='cp949')
df.head(3)
```

```python
df2 = df.rename(columns={'평균기온(°C)':'avg_temp', '최고기온(°C)':'max_temp'})
df2['일시'] = pd.to_datetime(df2['일시'], format='%Y-%m-%d')
df2.info()
```

```python
# 결측치 확인
print(df2.isnull().sum())

# 앞/뒤 값으로 보간(선형보간) — 기온처럼 연속적인 데이터에 적합
df2['avg_temp'] = df2['avg_temp'].interpolate(method='linear')
df2['max_temp'] = df2['max_temp'].interpolate(method='linear')

# 확인
print(df2.isnull().sum())
```

```python
df3 = df2.set_index('일시')
df_Dec = df3[pd.DatetimeIndex(df3.index).month == 12]
df_Dec
```

```python
plt.rcParams['figure.figsize'] = (8, 8)
plt.title('Jeju 2025. 12')
plt.plot(range(1, len(df_Dec)+1), df_Dec['max_temp'], label='max', c='r')
plt.plot(range(1, len(df_Dec)+1), df_Dec['avg_temp'], label='avg', c='y')

plt.xlim(1, 31)
plt.legend()
plt.show()
```
